```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv("HR.csv")


df.head()
```

```python

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nMissing values:")
print(df.isnull().sum())
```

```python

gender_count = df["Gender"].value_counts()

print("Employees by Gender:")
print(gender_count)


gender_count.plot(kind="bar", figsize=(8,5))

plt.title("Employee Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Employees")
plt.xticks(rotation=0)

plt.show()
```

```python

race_count = df["Race"].value_counts()

print("Employees by Race:")
print(race_count)


race_count.plot(kind="bar", figsize=(10,6))

plt.title("Employee Distribution by Race")
plt.xlabel("Race")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45, ha="right")

plt.show()
```

```python

gender_status = pd.crosstab(
    df["Gender"],
    df["Employement Status"]
)

print("Employment Status by Gender:")
print(gender_status)


gender_status.plot(kind="bar", figsize=(8,5))

plt.title("Employment Status by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Employees")
plt.xticks(rotation=0)

plt.show()
```

```python
gender_status_percentage = pd.crosstab(
    df["Gender"],
    df["Employement Status"],
    normalize="index"
) * 100

print("Employment Status Percentage by Gender:")
print(gender_status_percentage.round(2))
```

```python

race_status = pd.crosstab(
    df["Race"],
    df["Employement Status"]
)

print("Employment Status by Race:")
print(race_status)
```

```python

race_status_percentage = pd.crosstab(
    df["Race"],
    df["Employement Status"],
    normalize="index"
) * 100

print("Employment Status Percentage by Race:")
print(race_status_percentage.round(2))
```

```python


gender_status.plot(
    kind="bar",
    stacked=True,
    figsize=(9,6)
)

plt.title("Employment Status by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Employees")
plt.xticks(rotation=0)
plt.legend(title="Employment Status")
plt.show()
```

```python
# Overall HR dataset summary

print("Total employees:", len(df))
print("Number of departments:", df["Department"].nunique())
print("Departments:")
print(df["Department"].value_counts())

print("\nGender distribution:")
print(df["Gender"].value_counts())

print("\nRace distribution:")
print(df["Race"].value_counts())
```

```python


status_count = df["Employement Status"].value_counts()

print("Overall Employment Status:")
print(status_count)


status_percentage = df["Employement Status"].value_counts(normalize=True) * 100

print("\nOverall Employment Status Percentage:")
print(status_percentage.round(2))

status_count.plot(kind="bar", figsize=(8,5))

plt.title("Overall Employment Status")
plt.xlabel("Employment Status")
plt.ylabel("Number of Employees")
plt.xticks(rotation=0)

plt.show()
```
